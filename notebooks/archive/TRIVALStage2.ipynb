## Second way

This is another way to implement the BART model using torch package. The user can safely ignore the below code chunk and its output (unless they want to familarize themselves with different ways).

Manual PyTorch implementation from https://huggingface.co/facebook/bart-large-mnli

The user can directly move to next section i.e. "classification using SBERT".

```python
#import torch
from torch import cuda
#device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
device = f'cuda:{cuda.current_device()}' if cuda.is_available() else 'cpu';
print(device)
```

```python
# pose sequence as a NLI premise and label as a hypothesis
from transformers import AutoModelForSequenceClassification, AutoTokenizer
nli_model = AutoModelForSequenceClassification.from_pretrained('facebook/bart-large-mnli')
tokenizer = AutoTokenizer.from_pretrained('facebook/bart-large-mnli')

# Move the model to the correct device
nli_model = nli_model.to(device) # This line is added to move the model to the device

#sequence_to_classify = "one day I will see the world"
#label = ['travel', 'cooking', 'dancing', 'exploration']

premise = sequence_to_classify
label_probs = {}  # Store label probabilities
hypothesis = f'This example is {candidate_labels}.'
```

```python
for label in candidate_labels:
    hypothesis = f'This example is about {label}.'
    x = tokenizer.encode(premise, hypothesis, return_tensors='pt')
    logits = nli_model(x.to(device))[0]

    entail_contradiction_logits = logits[:, [0, 2]]
    probs = entail_contradiction_logits.softmax(dim=1)
    prob_label_is_true = probs[:, 1].item()  # Get probability as a number

    label_probs[label] = prob_label_is_true  # Store probability

```

```python
# Print or use the label_probs dictionary
for label, prob in label_probs.items():
    print(f"Label: {label}, Probability: {prob:.4f}")
```

Sorted based on probability

```python
sorted_label_probs = sorted(label_probs.items(), key=lambda item: item[1], reverse=True)
for label, prob in sorted_label_probs:
    print(f"Label: {label}, Probability: {prob:.4f}")
```

## Classification using SBERT

Earlier we used BART for ZSL. Here we use SBERT for scoring the text and different classes. Intutively, think of this model as another individual who in parallel is trying to validate the result.

The outcome from BART and SBERT can then we used to come to a consensus (expert judgement, intercoder reliability).

This approach uses sentence bert

```python
# load the sentence-bert model from the HuggingFace model hub
#!pip install transformers
from transformers import AutoTokenizer, AutoModel
from torch.nn import functional as F
tokenizer = AutoTokenizer.from_pretrained('deepset/sentence_bert')
model = AutoModel.from_pretrained('deepset/sentence_bert')

sentence = sequence_to_classify
labels = candidate_labels
```

```python
# run inputs through model and mean-pool over the sequence
# dimension to get sequence-level representations
inputs = tokenizer.batch_encode_plus([sentence] + labels,
                                     return_tensors='pt',
                                     padding=True
                                     )
input_ids = inputs['input_ids']
attention_mask = inputs['attention_mask']
output = model(input_ids, attention_mask=attention_mask)[0]
sentence_rep = output[:1].mean(dim=1)
label_reps = output[1:].mean(dim=1)

# now find the labels with the highest cosine similarities to
# the sentence
similarities = F.cosine_similarity(sentence_rep, label_reps)
closest = similarities.argsort(descending=True)
for ind in closest:
    print(f'label: {labels[ind]} \t similarity: {similarities[ind]}')
```
